```python
def Doesitmatch(n, arr, key):
  for i in range(n):
    if arr[i]== key:
      return i
    else:
      return -1

n = int(input("Give length of array: "))
arr = []
for i in range(n):
 arr.append(int(input(f"Give the {i+1}st term of array: ")))
key = int(input("Give the key: "))
result = Doesitmatch(n, arr, key)
if result == -1:
  print("sorry key not found")
else:
  print("key found in result")
```

```python
def reversearray(arr):
 n = len(arr)
 temp = [0]*n
 for i in range(n):
  temp[i] = arr[n-i-1]
 for i in range(n):
  arr[i] = temp[i]

arr = list(map(int, input("Please give array: ").split()))
reversearray(arr)
print(arr)
```
